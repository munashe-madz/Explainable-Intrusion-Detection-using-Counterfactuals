## Preprocessing

```python
df.nunique().sort_values()
```

#### Drop Identifier Features

```python
df.drop(columns=['Switch ID', 'Port Number', 'Table ID'], inplace=True)
```

#### Drop Dropped and Error Features
These features have constant values of 0.0

```python
df.drop(columns=['Packets Rx Dropped', 'Packets Tx Dropped', 'Packets Rx Errors', 'Packets Tx Errors'], inplace=True)
```

```python
df.drop(columns=['Delta Packets Rx Dropped', ' Delta Packets Tx Dropped', 'Delta Packets Rx Errors', 'Delta Packets Tx Errors'], inplace=True)
```

```python
df.describe(include='all')
```

```python
df.shape
```

Columns reduced from 34 to 23

#### Turning Binary Label to 1 and 0
Using map instead of get_dummies because we want to specify which label is 1

```python
df['Binary Label'] = df['Binary Label'].map({
    'Attack': 1,
    'Normal': 0
})
```

Let's determine how separable the binary classes are in terms of received packets and received bytes

```python
import numpy as np
df5 = df.copy()
fig, axes = plt.subplots(2, 2, figsize=(15, 10))
df5['Received Packets (log)'] = np.log1p(df['Received Packets'])
df5['Received Bytes (log)'] = np.log1p(df['Received Bytes'])
sns.histplot(data=df5, x='Received Packets (log)', hue='Binary Label', kde=True, ax=axes[0][0], stat='density', common_norm=False)
sns.histplot(data=df5, x='Received Bytes (log)', hue='Binary Label', kde=True, ax=axes[0][1], stat='density', common_norm=False)

sns.histplot(data=df5, x='Received Packets (log)', hue='Label', kde=True, ax=axes[1][0], stat='density', common_norm=False)
sns.histplot(data=df5, x='Received Bytes (log)', hue='Label', kde=True, ax=axes[1][1], stat='density', common_norm=False)
```

```python
df5['Avg Received Pkt Size'] = df5['Received Bytes'] / df5['Received Packets'].replace(0, np.nan)
sns.ecdfplot(data=df5, x=np.log1p(df5['Avg Received Pkt Size']), hue='Label')
```

```python
g = sns.displot(df5, x='Received Packets (log)', col='Label', col_wrap=3,
                kind='kde', fill=True, height=2.5)
```

```python
sns.boxplot(x='Binary Label', y='Packets Looked Up', data=df)
```

### Manually Label encoding the Multi-Class Labels

```python
df['Label'] = df['Label'].replace('Normal', 0)
df['Label'] = df['Label'].replace('TCP-SYN', 1)
df['Label'] = df['Label'].replace('Blackhole', 2)
df['Label'] = df['Label'].replace('Diversion', 3)
df['Label'] = df['Label'].replace('Overflow', 4)
df['Label'] = df['Label'].replace('PortScan', 5)
```

```python
df.to_csv('unr-multi-processed.csv')
```
